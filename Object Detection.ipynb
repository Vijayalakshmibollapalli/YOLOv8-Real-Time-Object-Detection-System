## Import Libraries

```python
import cv2
import time
from ultralytics import YOLO
from IPython.display import display, clear_output
from PIL import Image
import numpy as np
import PIL.Image
```

## Load YOLO Model

```python
# model = YOLO("yolov8s.pt")
model = YOLO("yolov8n.pt")  # faster, smoother
print("YOLO working")
```

## YOLO Webcam Code

```python
# Start webcam
cap = cv2.VideoCapture(0)

# Create fullscreen window
cv2.namedWindow("YOLO AI Detection System", cv2.WINDOW_NORMAL)
cv2.setWindowProperty("YOLO AI Detection System", cv2.WND_PROP_FULLSCREEN, cv2.WINDOW_FULLSCREEN)

prev_time = 0
record = False
video_writer = None

# Use fixed FPS for proper video playback
fixed_fps = 10   # adjust between 8–12 if needed

while True:

    # Read frame
    ret, frame = cap.read()
    if not ret:
        break

    # YOLO detection + tracking
    results = model.track(frame, persist=True, conf=0.50)

    # Draw detection boxes
    frame = results[0].plot()

    # Object counting
    object_count = {}
    
    if results[0].boxes is not None:
        classes = results[0].boxes.cls.tolist()
        names = model.names

        for cls in classes:
            label = names[int(cls)]
            object_count[label] = object_count.get(label, 0) + 1

    # FPS calculation
    curr_time = time.time()
    fps = 1/(curr_time-prev_time) if prev_time != 0 else 0
    prev_time = curr_time

    # Display FPS
    cv2.putText(frame,
                f"FPS: {int(fps)}",
                (20,40),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0,255,0),
                2)

    # Display object counts
    y_offset = 80
    for obj, count in object_count.items():
        cv2.putText(frame,
                    f"{obj}: {count}",
                    (20,y_offset),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    (255,0,0),
                    2)
        y_offset += 30
    
    # Video recording
    if record:
        if video_writer is None:
            fourcc = cv2.VideoWriter_fourcc(*'XVID')
            video_writer = cv2.VideoWriter(
                "output.avi",
                fourcc,
                fixed_fps,
                (frame.shape[1], frame.shape[0])
            )
        video_writer.write(frame)

    # Show frame in fullscreen
    cv2.imshow("YOLO AI Detection System", frame)

    # Keyboard controls
    key = cv2.waitKey(1) & 0xFF

    # Quit
    if key == ord('q'):
        break

    # Screenshot
    if key == ord('s'):
        filename = f"screenshot_{int(time.time())}.jpg"
        cv2.imwrite(filename, frame)
        print("Screenshot saved:", filename)

    # Toggle recording
    if key == ord('r'):
        record = not record
        print("Recording:", record)

        # Release writer when stopping recording
        if not record and video_writer is not None:
            video_writer.release()
            video_writer = None

# Release camera
cap.release()

if video_writer:
    video_writer.release()

cv2.destroyAllWindows()
```
